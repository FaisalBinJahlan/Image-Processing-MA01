Spatial Filtering – Smoothing and Sharpening

```
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

Task 1: Box Filter 7x7

```
image = cv2.imread('image.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

smoothed = cv2.blur(image, (7, 7))

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(smoothed)
plt.title("Smoothed (7x7 Box)")
plt.axis("off")

plt.show()
```

Task 2: Gaussian Filters

```
gaussian_5 = cv2.GaussianBlur(image, (5,5), 0)
gaussian_21 = cv2.GaussianBlur(image, (21,21), 0)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gaussian_5)
plt.title("Gaussian 5x5")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(gaussian_21)
plt.title("Gaussian 21x21")
plt.axis("off")

plt.show()
```

Task 3: Laplacian Sharpening

```
image_float = image.astype(np.float32) / 255.0

laplacian = cv2.Laplacian(image_float, cv2.CV_32F)

sharpened = np.clip(image_float - laplacian, 0, 1)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(sharpened)
plt.title("Sharpened")
plt.axis("off")

plt.show()
```
