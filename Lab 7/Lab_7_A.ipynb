# **Lab 7(A) - Laplacian Filter**

```python
import cv2
import numpy as np

from IPython.display import Image, display
from matplotlib import pyplot as plt

plt.rc('text', usetex=False)


```

## Apply Laplacian Filter on Image to get Sharp Image

```python
# ================ ( Laplacian Filter - Manual Kernel ) ===================

# 0. Image file path
imgFile = "moon.jpg"

# 1. Load image and convert to grayscale
img = cv2.imread(imgFile, 0)

if img is None:
    raise FileNotFoundError("moon.jpg was not found. Put moon.jpg in the same folder as this notebook.")

# 2. Manually define the Laplacian Kernel
manual_kernel = np.array([
    [0,  1,  0],
    [1, -4,  1],
    [0,  1,  0]
], dtype=np.float32)

# 3. Apply the filter using filter2D
laplacian_manual = cv2.filter2D(img, cv2.CV_64F, manual_kernel)

# 4. Convert Laplacian result to 8-bit for display
laplacian_abs = cv2.convertScaleAbs(laplacian_manual)

# 5. Add edge information to the original image to get sharper image
finalImage = cv2.add(img, laplacian_abs)

# 6. Display images
plt.figure(figsize=(12, 10))

plt.subplot(221)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(222)
plt.imshow(laplacian_manual, cmap='gray')
plt.title('Laplacian Image')
plt.axis('off')

plt.subplot(223)
plt.imshow(laplacian_abs, cmap='gray')
plt.title('Convert with Abs Scale')
plt.axis('off')

plt.subplot(224)
plt.imshow(finalImage, cmap='gray')
plt.title('Final Sharpened Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# ================== ( Laplacian Filter - OpenCV Function ) ===============

# 1. Optional: Apply Gaussian Blur to reduce noise
blurred = cv2.GaussianBlur(img, (3, 3), 0)

# 2. Apply the Laplacian Algorithm
laplacian = cv2.Laplacian(blurred, cv2.CV_64F)

# 3. Convert back to 8-bit image to view it properly
laplacian_8bit = cv2.convertScaleAbs(laplacian)

# 4. Add Laplacian edges to original image
finalImage2 = cv2.add(img, laplacian_8bit)

# 5. Display images
plt.figure(figsize=(12, 10))

plt.subplot(221)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(222)
plt.imshow(laplacian, cmap='gray')
plt.title('Laplacian Image')
plt.axis('off')

plt.subplot(223)
plt.imshow(laplacian_8bit, cmap='gray')
plt.title('Convert with Abs Scale')
plt.axis('off')

plt.subplot(224)
plt.imshow(finalImage2, cmap='gray')
plt.title('Final Sharpened Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```
